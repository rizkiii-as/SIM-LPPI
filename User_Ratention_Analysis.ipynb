# Data Cleansing

```python
df_clean = df.copy() #menyalin dataframe asli
df_clean['order_date'] = df_clean['order_date'].astype('datetime64[ns]') #konversi kolom order_date menjadi datetime
df_clean['year_month'] = df_clean['order_date'].dt.to_period('M') #membuat kolom year_month
df_clean = df_clean[~df_clean['customer_id'].isna()] #menghapus semua baris tanpa customer_id
df_clean = df_clean[~df_clean['product_name'].isna()]#menghapus semua baris tanpa product_name
df_clean['product_name'] = df_clean['product_name'].str.lower()#membuat product_name menjadi huruf kecil
df_clean = df_clean[(~df_clean['product_code'].str.lower().str.contains('test')) |
                    (~df_clean['product_name'].str.contains('test'))] #menghapus semua baris dengan product_code atau product_name test
df_clean['order_status'] = np.where(df_clean['order_id'].str[:1]=='C', 'cancelled', 'delivered') # membuat kolom order_status dengan nilai 'cancelled' jika order_id diawali dengan huruf 'c' dan 'delivered' jika order_id tanpa awalan huruf 'c'
df_clean['quantity'] = df_clean['quantity'].abs() # mengubah nilai quantity yang negatif menjadi positif karena nilai negatif tersebut hanya menandakan order tersebut cancelled
df_clean = df_clean[df_clean['price']>0] # menghapus baris dengan price bernilai negatif
df_clean['amount'] = df_clean['quantity'] * df_clean['price'] # membuat nilai amount, yaitu perkalian antara quantity dan price

# mengganti product_name dari product_code yang memiliki beberapa product_name dengan salah satu product_name-nya yang paling sering muncul
most_freq_product_name = df_clean.groupby(['product_code','product_name'], as_index=False).agg(order_cnt=('order_id','nunique')).sort_values(['product_code','order_cnt'], ascending=[True,False])
most_freq_product_name['rank'] = most_freq_product_name.groupby('product_code')['order_cnt'].rank(method='first', ascending=False)
most_freq_product_name = most_freq_product_name[most_freq_product_name['rank']==1].drop(columns=['order_cnt','rank'])
df_clean = df_clean.merge(most_freq_product_name.rename(columns={'product_name':'most_freq_product_name'}), how='left', on='product_code')
df_clean['product_name'] = df_clean['most_freq_product_name']
df_clean = df_clean.drop(columns='most_freq_product_name')

df_clean['customer_id'] = df_clean['customer_id'].astype(str) # mengkonversi customer_id menjadi string

# menghapus outlier
from scipy import stats
df_clean = df_clean[(np.abs(stats.zscore(df_clean[['quantity','amount']]))<3).all(axis=1)]
df_clean = df_clean.reset_index(drop=True)
df_clean
```

```python
df_clean.info()
```

# Membuat User Retention Cohort

## Agregat Data Transaksi Ke Bentuk Summary Total Transaksi/Order Setiap Pengguna Setiap Bulan

```python
df_user_monthly = df_clean.groupby(['customer_id', 'year_month'], as_index=False).agg(order_cnt=('order_id', 'nunique'))
df_user_monthly
```

## Buat Kolom Sebagai Cohort dari Pengguna, Misal Cohort Bulan Pertama Kali Bertransaksi

```python
df_user_monthly['cohort'] = df_user_monthly.groupby('customer_id')['year_month'].transform('min')
df_user_monthly
```

## Menghitung Jarak Bulan Antara Bulan Transaksi dengan Bulan Pertama Kali Transaksi dan Jumlahkan dengan 1 agar Jarak Bulan 0 Menjadi 1 yang Berarti Bulan Pertama, dst

```python
df_user_monthly['period_num'] = (df_user_monthly['year_month'] - df_user_monthly['cohort']).apply(attrgetter('n')) + 1
df_user_monthly
```

## Tabel pivot dengan index berupa cohort, kolom berupa jarak bulan, dan nilainya adalah banyaknya pengguna unik (count unique dari ID pengguna)

```python
df_cohort_pivot = pd.pivot_table(df_user_monthly, index='cohort', columns='period_num', values='customer_id', aggfunc=pd.Series.nunique)
df_cohort_pivot
```

## Hitung banyaknya pengguna di masing-masing cohort (pengguna yang pertama kali transaksi di bulan tersebut) dan bagi semua nilai di tabel pivot tadi dengan nilai tersebut sebagai retention rate

```python
cohort_size = df_cohort_pivot.iloc[:, 0]
cohort_size
```

```python
df_retention_cohort = df_cohort_pivot.divide(cohort_size, axis=0)
df_retention_cohort
```

## Tampilkan tabel pivot yang sudah berisi nilai retention rate tersebut dalam heatmap

```python
with sns.axes_style('white'):

    fig, ax = plt.subplots(1, 2, figsize=(12, 8), sharey=True, gridspec_kw={'width_ratios':[1, 11]})

    # user retention cohort
    sns.heatmap(df_retention_cohort, annot=True, fmt='.0%', cmap='Blues', ax=ax[1])
    ax[1].set_title('User Retention Cohort')
    ax[1].set(xlabel='Month Number', ylabel='')

    # cohort size
    df_cohort_size = pd.DataFrame(cohort_size)
    white_cmap = mcolors.ListedColormap(['white'])
    sns.heatmap(df_cohort_size, annot=True, cbar=False, fmt='g', cmap=white_cmap, ax=ax[0])
    ax[0].tick_params(bottom=False)
    ax[0].set(xlabel='Cohort Size', ylabel='First Order Month', xticklabels=[])

    fig.tight_layout()
```
